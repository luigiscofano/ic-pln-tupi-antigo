# Auditoria reproduzível

Notebook novo de demonstração com dados sintéticos. Objetivos: preservar formas originais, detectar incompletude sem inventar traduções e separar métricas de qualidade estrutural de validação linguística.

```python
from pathlib import Path
import sys
root=Path.cwd()
if not (root/'tupi').exists(): root=root.parent
sys.path.insert(0,str(root))
from tupi import importar_tsv,auditar,cer,wer
registros=importar_tsv(root/'data/demo.tsv')
assert len(registros)==3
registros
```

## Revisão humana
Dois registros homógrafos não são eliminados. A tradução ausente permanece vazia. NFC não implica ortografia correta.

```python
relatorio=auditar(registros)
assert relatorio['flagged_records']==2
assert not relatorio['linguistic_validation']
relatorio
```

## Métricas ilustrativas
Estes pares são artificiais. Não representam taxa de erro do dicionário ou de um mecanismo de OCR.

```python
assert cer('abc','axc')==1/3
assert wer('a b','a x b')==0.5
{'CER_sintetico':cer('abc','axc'),'WER_sintetico':wer('a b','a x b')}
```
